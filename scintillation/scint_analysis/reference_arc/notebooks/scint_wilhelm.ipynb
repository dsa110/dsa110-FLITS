```python
from scipy.fft import fft, fftshift
import numpy as np
from scipy.optimize import curve_fit
from tqdm import tqdm
import matplotlib.pyplot as plt
import scipy.constants as cons

# --- Begin updated functions with NaN handling ---

def normalize_spectrum(on_spec, off_spec):
    """
    Subtract off-pulse mean and divide by off-pulse rms to yield per-channel S/N,
    properly ignoring NaNs.
    """
    off_mean = np.nanmean(off_spec, axis=0)
    off_rms  = np.nanstd(off_spec, axis=0)
    if np.isscalar(off_rms):
        off_rms = np.array(off_rms)
    # avoid zero division
    off_rms = np.where(off_rms == 0, np.nan, off_rms)
    return (on_spec - off_mean) / off_rms


def compute_acf(spec):
    """
    Compute normalized spectral autocorrelation of a 1D spectrum, handling NaNs.
    Returns array of length Nlags = len(spec).
    """
    spec = np.atleast_1d(spec)
    N = spec.size
    if N < 2:
        return np.full(N, np.nan)
    valid = ~np.isnan(spec)
    mean_spec = np.nanmean(spec)
    centered = spec - mean_spec
    acf = np.full(N, np.nan)
    for lag in range(N):
        i1 = np.arange(N - lag)
        i2 = i1 + lag
        mask = valid[i1] & valid[i2]
        if not np.any(mask):
            continue
        num = np.sum(centered[i1[mask]] * centered[i2[mask]])
        denom = mask.sum() * (mean_spec - 1.0)
        acf[lag] = num / denom if denom != 0 else np.nan
    return acf


def multi_lorentzian(x, *params):
    """
    Sum of Lorentzians: params = [m1, g1, m2, g2, ...]
    Each term = m^2 / (1 + (x/g)**2)
    """
    y = np.zeros_like(x)
    for m, g in zip(params[0::2], params[1::2]):
        y += (m**2) / (1 + (x / g)**2)
    return y


def fit_acf(lags, acf, n_screens=2):
    """
    Fit (1 + n_screens) Lorentzians to ACF vs lags. Return list of (m,g).
    """
    mask = (~np.isnan(acf)) & (lags != 0)
    x = lags[mask]
    y = acf[mask]
    if x.size < 2*(n_screens+1):
        return []
    # initial guesses: one broad + n_screens narrow
    p0 = [0.5, lags.max()*0.5]
    for _ in range(n_screens): p0 += [0.8, lags.max()*0.1]
    try:
        popt, _ = curve_fit(multi_lorentzian, x, y, p0=p0, maxfev=10000)
    except Exception:
        return []
    ms = popt[0::2]
    gs = popt[1::2]
    return list(zip(ms, gs))


def measure_scintillation(freq, dynspec, on_idx, off_idx, n_subbands=8):
    """
    Measure full-band and sub-band scintillation: returns full_fits, sub_fits_list.
    dynspec: 2D array (ntime,nfreq) or (nfreq,ntime).
    on_idx,off_idx: index arrays for on/off pulses along time axis.
    """
    dynspec = np.asarray(dynspec)
    if dynspec.ndim != 2:
        raise ValueError(f"dynspec must be 2D, got {dynspec.shape}")
    # ensure shape (ntime,nfreq)
    if dynspec.shape[1] != freq.size and dynspec.shape[0] == freq.size:
        dynspec = dynspec.T
    if dynspec.shape[1] != freq.size:
        raise ValueError(f"freq length {freq.size} != dynspec second dim {dynspec.shape[1]}")

    # time-average on/off
    on_spec  = np.nanmean(dynspec[on_idx], axis=0)
    off_spec = np.nanmean(dynspec[off_idx], axis=0)
    norm_spec = normalize_spectrum(on_spec, off_spec)

    # full-band
    acf_full = compute_acf(norm_spec)
    df = freq[1]-freq[0] if freq.size>1 else 1.0
    lags_full = np.arange(acf_full.size)*df
    full_fits = fit_acf(lags_full, acf_full, n_screens=2)

    # sub-bands equal S/N
    valid = ~np.isnan(norm_spec)
    cum_sn = np.nancumsum(np.where(valid, norm_spec, 0))
    if cum_sn[-1] <= 0:
        return full_fits, []
    idx = np.arange(norm_spec.size)
    edges = np.interp(np.linspace(0, cum_sn[-1], n_subbands+1), cum_sn, idx)

    sub_results = []
    for i in range(n_subbands):
        lo, hi = int(edges[i]), int(edges[i+1])
        if hi<=lo: sub_results.append([]); continue
        sub = norm_spec[lo:hi]
        if np.all(np.isnan(sub)): sub_results.append([]); continue
        acf_sub = compute_acf(sub)
        lags_sub = np.arange(acf_sub.size)*df
        sub_results.append(fit_acf(lags_sub, acf_sub, n_screens=1))

    return norm_spec, full_fits, sub_results

# --- Visualization helpers ---
def plot_subband_acfs(freq, norm_spec, sub_results, df, n_subbands=8):
    """
    Plot ACFs and Lorentzian fits for each sub-band.
    freq: full freq axis
    norm_spec: normalized spectrum
    sub_results: output from measure_scintillation[1]
    df: channel spacing
    """
    # compute sub-band freq centers and index edges
    valid = ~np.isnan(norm_spec)
    cum_sn = np.nancumsum(np.where(valid, norm_spec, 0))
    idx = np.arange(norm_spec.size)
    edges = np.interp(np.linspace(0, cum_sn[-1], n_subbands+1), cum_sn, idx)
    centers = []
    for i in range(n_subbands):
        lo, hi = int(edges[i]), int(edges[i+1])
        centers.append(freq[lo:hi].mean() if hi>lo else np.nan)

    # plot
    plt.figure(figsize=(8,6))
    for i, center in enumerate(centers):
        lo, hi = int(edges[i]), int(edges[i+1])
        sub = norm_spec[lo:hi]
        acf_sub = compute_acf(sub)
        lags = np.arange(acf_sub.size)*df
        plt.plot(lags, acf_sub + i*1.2, label=f'{center:.1f} MHz')
        fit = sub_results[i]
        if fit:
            yfit = multi_lorentzian(lags, fit[0][0], fit[0][1])
            plt.plot(lags, yfit + i*1.2, '--')
    plt.xlabel('Frequency lag (MHz)')
    plt.ylabel('ACF + offset')
    plt.legend()
    plt.title('Sub-band ACFs and Lorentzian fits')
    plt.show()

    return centers

def plot_scintillation_law(centers, sub_results):
    """
    Plot decorrelation bandwidth vs sub-band center and fit power law.
    """
    # extract first gamma from each subband
    gammas = [res[0][1] if res else np.nan for res in sub_results]
    centers = np.array(centers)
    gammas = np.array(gammas)
    mask = ~np.isnan(gammas)
    p = np.polyfit(np.log(centers[mask]), np.log(gammas[mask]), 1)
    alpha, logA = p
    plt.figure()
    plt.loglog(centers[mask], gammas[mask], 'o', label='data')
    x = np.linspace(centers.min(), centers.max(), 100)
    plt.loglog(x, np.exp(logA)*x**alpha, '-', label=f'fit α={alpha:.2f}')
    plt.xlabel('Sub-band center (MHz)')
    plt.ylabel('Decorrelation bandwidth (MHz)')
    plt.legend()
    plt.title('Scintillation bandwidth vs frequency')
    plt.show()
```
